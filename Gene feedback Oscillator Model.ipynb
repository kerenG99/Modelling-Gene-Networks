```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import odeint 
```

```python
#setting initial condition
y0 = [0,0,0] #G1, G2, G3

t= np.linspace(0,300, num=150) #from o to 300 with 150 timepoints

#define parameters
k_1 =0.5
gamma_1 = 0.1 #should be smaller degradation rate
k_2 = 0.5
gamma_2 = 0.1
k_3 = 0.5
gamma_3 = 0.1
n = 9
c = 1

params = [k_1, gamma_1,k_2,gamma_2,k_3,gamma_3,n,c] #array



```

```python
def negOS(variables,t,params) :
    G1 = variables[0]
    G2 = variables[1]
    G3 = variables[2]

    k_1 = params [0]
    gamma_1 = params [1]
    k_2 = params [2]
    gamma_2 = params [3]
    k_3 = params [4]
    gamma_3 = params [5]
    n = params [6]
    c = params [7]
    

    dG1dt = (c**n/(c**n + G3**n))* k_1 - gamma_1 * G1
    dG2dt = (G1**n/(c**n + G1**n))* k_2 - gamma_2 * G2
    dG3dt = (G2**n/(c**n + G2**n))* k_3 - gamma_3 * G3

    return ([dG1dt,dG2dt, dG3dt]) 
```

```python
y = odeint(negOS, y0, t, args=(params,))
```

```python
f, (ax1, ax2, ax3) = plt.subplots(3, sharex = True, sharey = False)

line1, = ax1.plot(t,y[:,0], color="b", label="G1")
line2, = ax2.plot(t,y[:,1], color="r", label="G2")
line3, = ax3.plot(t,y[:,2], color="g", label="G3")

y = odeint(negOS, y0, t, args=(params,))
ax1.set_ylabel("Number")
ax2.set_xlabel("Time")

ax1.legend(handles=[line1,line2,line3])

plt.show()
            
```
